```python
def get_bats(inp):
    arr = []

    for c in inp:

        arr.append(int(c))

        # search through first through second last element

        max = 0

    position = 0

    for i in range(len(arr) - 1):

        if arr[i] > max:

            max = arr[i]

            position = i

    sec = 0

    # get second digit

    for j in arr[position+1:]:

        if j > sec:

            sec = j

        

    bats = str(max) + str(sec)

    return int(bats)
```

```python
# sliding window function rather than what the problem needed 
def get_bats_large(inp):
    window_size = 12
    max = 0 

    for j in range(len(inp) - window_size):
        window = inp[j:j+12]
        #print('current window' + str(inp[j:j+12]))
        if int(window) > max:
            # print('updating max to' + window)
            max = int(window)


    return max
```

```python
def get_joltage(inp):
    arr = list(inp)
    joltage = ''
    start = 0
    need = 12

    while need > 0:
        leave = need - 1
        end = len(arr) - leave

        window = arr[start:end]
        max_value = max(window)

        index_of_max = start + window.index(max_value)

        joltage += max_value
        start = index_of_max + 1
        need -= 1
        
    return int(joltage)
```

```python
total3 = 0
for bank in input:
    total3 += get_joltage(bank)
```
