```python
with open('Day6Input.txt', 'r') as file:
    inp = file.read()
```

```python
split = inp.splitlines()

full_list = [line.split() for line in split]
```

```python
# break out into operator and numbers lists 
numbers = full_list[:-1]
operators = full_list[-1]
```

```python
operations = []
for i in range(len(numbers[0])):
    operations.append(numbers[0][i] + operators[i] + numbers[1][i] + operators[i] + numbers[2][i] + operators[i] + numbers[3][i])
```

```python
solved = [eval(i) for i in operations]
```

```python
sum(solved)
```
